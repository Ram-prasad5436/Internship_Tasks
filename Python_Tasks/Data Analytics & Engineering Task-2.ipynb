### Problem-Solving Approach
    Follow these six steps for every question:
    Observe → Understand → Break → Logic → Code → Test
    Observe: Identify the given input and expected output.
    Understand: Determine exactly what the problem is asking.
    Break: Divide the problem into smaller, manageable steps.
    Logic: Develop the solution approach in simple English or pseudocode.
    Code: Implement your logic using Python.
    Test: Execute the code and verify the output with different inputs.

### Q1. Find the Shortest Word in a Sentence
    Write a Python program to find the shortest word in a given sentence.
    Example input:
    Python makes coding easy
    Expected output:
    easy


```python
a = 'python makes coding easy'
words = a.split()
short = words[0]
for ch in words:
    if len(short)>len(ch):
        short = ch
print(short)
```

### Q2. Find the Largest Number in a List
    Write a Python program to find the largest number in a list without using the built-in max() function.

```python
a=[12,45,89,23]
largest=a[0]
for i in a:
    if largest < i:
        largest=i
print(largest)
    
```

### Q6. Count Vowels in a String
    Write a Python program to count the total number of vowels in a string. Consider both uppercase and lowercase vowels.
    Example input:
    Education
    Expected output:
    5


```python
a='Education'
vowels = 0
b=a.lower()
for ch in b:
    if ch in 'aeiou':
        vowels += 1
print(vowels)
```

### Q7. Count Consonants in a String
    Write a Python program to count the consonants in a string. Ignore spaces, digits, and special characters.
    Example input:
    Python 123!
    Expected output:
    5


```python
a = 'Python 123!'
b = a.strip()
c=b.lower()
consonants=0
for ch in c:
    if ch.isalpha() and ch not in 'aeiou':
        consonants+=1
print(consonants)
```

### Q8. Find the Most Frequent Character
    Write a Python program to find the character that appears most frequently in a string. If multiple characters have the same highest frequency, return the first one encountered. Ignore spaces.
    Example input:
    banana
    Expected output:
    a


```python
a='banana'
count = {}
for ch in a:
    if ch != ' ':
        count[ch] = count.get(ch,0)+1
result = max(count,key=count.get)
print(result)
```

### Q10. Find All Characters Appearing More Than Once

    Example input:
    programming
    Expected output:
    ['r', 'g', 'm']


```python
s = 'programming'
count = {}
for ch in s:
    if ch != ' ':
        count[ch] = count.get(ch,0)+1
result = []
for ch in count:
    if count[ch]>1:
        result.append(ch)
print(result)
```

### Q15. Create a Dictionary of Numbers and Their Even/Odd Status
    Write a Python program to create a dictionary containing each number as a key and "Even" or "Odd" as its value.
    Example input:
    [1, 2, 3, 4, 5]
    Expected output:
    {1: 'Odd', 2: 'Even', 3: 'Odd', 4: 'Even', 5: 'Odd'}


```python
r = [1,2,3,4,5]
result ={}
for i in r:
    if i % 2 == 0:
        result[i]= 'Even'
    else:
        result[i]= 'Odd'
print(result)
```

### Q16. Find the Most Frequent Word in a Sentence
    Write a Python program to find the word that appears most frequently in a sentence. Treat uppercase and lowercase versions of a word as the same word. If multiple words share the highest frequency, return the one that appears first.
