## Q1. Find the Frequency of Each Word in a Sentence

### Pseudocode
1. Create an empty dictionary.
2. Split the sentence into individual words.
3. Count how many times each word appears.
4. Return the dictionary containing each word and its frequency.

### Expected Output
{'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'powerful': 1}


```python

# Q1. Find the Frequency of Each Word in a Sentence

def q1_word_frequency(sentence):
    counts = {}
    for word in sentence.split():
        counts[word] = counts.get(word, 0) + 1
    return counts

# Assignment example
print(q1_word_frequency(
    "python is easy and python is powerful"
))

# Additional test case
print(q1_word_frequency("red blue red"))
```


## Q2. Find the Frequency of Each Character in a Word

### Pseudocode
1. Create an empty  dictionary.
2.Read each character in the given word.
3. If the character exists in the dictionary, increase its count by 1.
4. Otherwise, add the character to the dictionary with a count of 1.
5.Return the dictionary.

### Example Input:
banana

### Expected Output:
{'b': 1, 'a': 3, 'n': 2}


```python

# Q2. Find the Frequency of Each Character in a Word

def q2_character_frequency(word):
    counts = {}

    for char in word:
        if char in counts:
            counts[char] += 1
        else:
            counts[char] = 1

    return counts


# Assignment example
print(q2_character_frequency("banana"))

# Additional test case
print(q2_character_frequency("hello"))
```


## Q3. Find the First Non-Repeating Character in a Word

### Pseudocode
1. Count the frequency of every character in the word.
2. Traverse the word from left to right.
3. Check the frequency of each character.
4. Return the first character whose frequency is 1.
5. If no such character exists, return None.

### Example Input
swiss

### Expected Output
w


```python

# Q3. Find the First Non-Repeating Character in a Word

def q3_first_non_repeating(word):
    counts = {}

    for char in word:
        counts[char] = counts.get(char, 0) + 1

    for char in word:
        if counts[char] == 1:
            return char

    return None


# Assignment example
print(q3_first_non_repeating("swiss"))

# Additional test case
print(q3_first_non_repeating("aabb"))
```


## Q4. Check Whether Two Strings Are Anagrams

### Pseudocode
1. Take two input strings.
2. Count the frequency of each character in both strings.
3. Compare the two frequency dictionaries.
4. If both dictionaries are equal, return True.
5. Otherwise, return False.

### Example Input
s1 = "listen"
s2 = "silent"

### Expected Output
True


```python

# Q4. Check Whether Two Strings Are Anagrams

def q4_are_anagrams(s1, s2):
    count1 = {}
    count2 = {}

    for char in s1:
        count1[char] = count1.get(char, 0) + 1

    for char in s2:
        count2[char] = count2.get(char, 0) + 1

    return count1 == count2


# Assignment example
print(q4_are_anagrams("listen", "silent"))

# Additional test case
print(q4_are_anagrams("hello", "world"))
```


## Q5. Find the First Repeating Character in a String

### Pseudocode
1. Create an empty set to store visited characters.
2. Read the string from left to right.
3. If the current character is already in the set, return that character.
4. Otherwise, add the character to the set.
5. If no character repeats, return None.

### Example Input
programming

### Expected Output
r


```python

# Q5. Find the First Repeating Character in a String

def q5_first_repeating(string):
    seen = set()

    for char in string:
        if char in seen:
            return char
        seen.add(char)

    return None


# Assignment example
print(q5_first_repeating("programming"))

# Additional test case
print(q5_first_repeating("abcdef"))
```


## Q6. Move All Zeros to the End of a List

### Pseudocode
1. Take a list of numbers.
2. Create a new list containing all non-zero elements in their original order.
3. Create another list containing all zero elements.
4. Combine the two lists.
5. Return the resulting list.

### Example Input
[0, 3, 0, 1, 2, 0, 4]

### Expected Output
[3, 1, 2, 4, 0, 0, 0]


```python

# Q6. Move All Zeros to the End of a List

def q6_move_zeros(nums):
    non_zeros = [num for num in nums if num != 0]
    zeros = [num for num in nums if num == 0]

    return non_zeros + zeros


# Assignment example
print(q6_move_zeros([0, 3, 0, 1, 2, 0, 4]))

# Additional test case
print(q6_move_zeros([0, 1, 0, 2]))
```


## Q7. Find the Second Highest Distinct Number in a List

### Pseudocode
1. Take a list of numbers.
2. Remove duplicate values.
3. Sort the distinct numbers in descending order.
4. If at least two distinct numbers exist, return the second number.
5. Otherwise, return None.

### Example Input
[10, 20, 40, 30, 40, 20]

### Expected Output
30


```python

# Q7. Find the Second Highest Distinct Number in a List

def q7_second_highest(nums):
    distinct_numbers = sorted(set(nums), reverse=True)

    if len(distinct_numbers) < 2:
        return None

    return distinct_numbers[1]


# Assignment example
print(q7_second_highest([10, 20, 40, 30, 40, 20]))

# Additional test case
print(q7_second_highest([5, 5, 5]))
```


## Q8. Remove Duplicates Without Changing the Original Order

### Pseudocode
1. Take a list of elements.
2. Create an empty result list and an empty set.
3. Traverse the input list from left to right.
4. If an element has not appeared before, add it to the result list and the set.
5. Skip elements that have already appeared.
6. Return the result list.

### Example Input
[4, 2, 4, 5, 2, 7, 5]

### Expected Output
[4, 2, 5, 7]


```python

# Q8. Remove Duplicates Without Changing the Original Order

def q8_remove_duplicates(nums):
    result = []
    seen = set()

    for num in nums:
        if num not in seen:
            result.append(num)
            seen.add(num)

    return result


# Assignment example
print(q8_remove_duplicates([4, 2, 4, 5, 2, 7, 5]))

# Additional test case
print(q8_remove_duplicates([1, 1, 2, 2, 3]))
```


## Q9. Find the Missing Number from 1 to N

### Pseudocode
1. Take a list of numbers and the value N.
2. Calculate the expected sum of numbers from 1 to N using N * (N + 1) / 2.
3. Calculate the sum of the numbers in the given list.
4. Subtract the actual sum from the expected sum.
5. Return the difference as the missing number.

### Example Input
nums = [1, 2, 4, 5, 6]
N = 6

### Expected Output
3


```python

# Q9. Find the Missing Number from 1 to N

def q9_missing_number(nums, n):
    expected_sum = n * (n + 1) // 2
    actual_sum = sum(nums)

    return expected_sum - actual_sum


# Assignment example
print(q9_missing_number([1, 2, 4, 5, 6], 6))

# Additional test case
print(q9_missing_number([1, 3, 4], 4))
```


## Q10. Find Two Numbers Whose Sum Equals a Target

### Pseudocode
1. Take a list of numbers and a target value.
2. Create an empty set to store previously visited numbers.
3. Traverse the list from left to right.
4. For each number, calculate the required number by subtracting it from the target.
5. If the required number is already in the set, return the pair.
6. Otherwise, add the current number to the set.
7. If no valid pair is found, return an empty list.

### Example Input
nums = [2, 7, 11, 15]
target = 9

### Expected Output
[2, 7]


```python

# Q10. Find Two Numbers Whose Sum Equals a Target

def q10_two_sum_first_pair(nums, target):
    seen = set()

    for num in nums:
        needed = target - num

        if needed in seen:
            return [needed, num]

        seen.add(num)

    return []


# Assignment example
print(q10_two_sum_first_pair([2, 7, 11, 15], 9))

# Additional test case
print(q10_two_sum_first_pair([1, 2, 3], 20))
```


## Q11. Find the Number Closest to Zero

### Pseudocode
1. Take a list of numbers.
2. If the list is empty, return None.
3. Compare the absolute values of the numbers.
4. Select the number with the smallest absolute value.
5. If two numbers are equally close to zero, choose the positive number.
6. Return the selected number.

### Example Input
[-8, -3, 2, 5, -2]

### Expected Output
2


```python

# Q11. Find the Number Closest to Zero

def q11_closest_to_zero(nums):
    if not nums:
        return None

    return min(nums, key=lambda num: (abs(num), num < 0))


# Assignment example
print(q11_closest_to_zero([-8, -3, 2, 5, -2]))

# Additional test case
print(q11_closest_to_zero([-4, 1, 3]))
```


## Q12. Find Common Elements Between Two Lists

### Pseudocode
1. Take two lists as input.
2. Convert the second list into a set for quick membership checking.
3. Create an empty result list and an empty set to track added elements.
4. Traverse the first list in its original order.
5. If an element exists in the second list and has not been added yet, append it to the result list.
6. Return the result list.

### Example Input
list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]

### Expected Output
[3, 4, 5]


```python

# Q12. Find Common Elements Between Two Lists

def q12_common_elements(list1, list2):
    values = set(list2)
    result = []
    added = set()

    for num in list1:
        if num in values and num not in added:
            result.append(num)
            added.add(num)

    return result


# Assignment example
print(q12_common_elements(
    [1, 2, 3, 4, 5],
    [3, 4, 5, 6, 7]
))

# Additional test case
print(q12_common_elements(
    [1, 2, 2, 3],
    [2, 3, 4]
))
```


## Q13. Rotate a List to the Right by K Positions

### Pseudocode
1. Take a list of numbers and the number of positions K.
2. If the list is empty, return an empty list.
3. Calculate K modulo the length of the list.
4. Take the last K elements and move them to the beginning.
5. Keep the remaining elements in their original order.
6. Return the rotated list.

### Example Input
nums = [1, 2, 3, 4, 5]
K = 2

### Expected Output
[4, 5, 1, 2, 3]


```python

# Q13. Rotate a List to the Right by K Positions

def q13_rotate_right(nums, k):
    if not nums:
        return []

    k = k % len(nums)

    if k == 0:
        return nums[:]

    return nums[-k:] + nums[:-k]


# Assignment example
print(q13_rotate_right([1, 2, 3, 4, 5], 2))

# Additional test case
print(q13_rotate_right([1, 2, 3], 0))
```


## Q14. Find the Maximum Sum of Two Consecutive Elements

### Pseudocode
1. Take a list containing at least two numbers.
2. Calculate the sum of the first two consecutive elements.
3. Move through the list and calculate the sum of each consecutive pair.
4. Keep track of the maximum sum found.
5. Return the maximum sum.

### Example Input
[4, 7, 2, 9, 5]

### Expected Output
14


```python

# Q14. Find the Maximum Sum of Two Consecutive Elements

def q14_max_consecutive_sum(nums):
    max_sum = nums[0] + nums[1]

    for i in range(1, len(nums) - 1):
        current_sum = nums[i] + nums[i + 1]

        if current_sum > max_sum:
            max_sum = current_sum

    return max_sum


# Assignment example
print(q14_max_consecutive_sum([4, 7, 2, 9, 5]))

# Additional test case
print(q14_max_consecutive_sum([1, 10, 2]))
```


## Q15. Find All Pairs of Numbers Whose Sum Equals a Target

### Pseudocode
1. Take a list of numbers and a target value.
2. Create an empty set to store unique pairs.
3. Check every pair of elements at different positions in the list.
4. If the sum of a pair equals the target, arrange the two numbers in ascending order.
5. Add the pair to the set to avoid duplicate pairs.
6. Sort the pairs by their first element.
7. Return the list of pairs.

### Example Input
nums = [2, 4, 3, 5, 7, 8, 1]
target = 9

### Expected Output
[[1, 8], [2, 7], [4, 5]]


```python

# Q15. Find All Pairs of Numbers Whose Sum Equals a Target

def q15_all_pairs(nums, target):
    pairs = set()

    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                pair = tuple(sorted((nums[i], nums[j])))
                pairs.add(pair)

    return [list(pair) for pair in sorted(pairs)]


# Assignment example
print(q15_all_pairs([2, 4, 3, 5, 7, 8, 1], 9))

# Additional test case
print(q15_all_pairs([1, 2, 3], 10))
```
